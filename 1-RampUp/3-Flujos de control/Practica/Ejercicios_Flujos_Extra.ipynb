* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción

```python
def sumar(a, b):
    return a + b

def restar(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def dividir(a, b):
    if b == 0:
        raise ValueError("No se puede dividir entre cero.")
    return a / b

def potencia(a, b):
    return a ** b

# Pedimos el número de forma robusta
def pedir_numero(mensaje):
    while True:
        try:
            return float(input(mensaje))
        except ValueError:
            print("❌ Error: Debes introducir un valor numérico válido. Inténtalo de nuevo.")

# Función principal de la calculadora
def calculadora():
    historial = []  # Aquí guardo el historial

    print("=== CALCULADORA INTERACTIVA ===")
    while True:
        print("\n¿Qué operación deseas realizar?")
        print(" - suma")
        print(" - resta")
        print(" - multiplicación")
        print(" - división")
        print(" - potencia")
        print(" - historial")
        print(" - salir")

        opcion = input("Elige una opción: ").lower().strip()

        # Condición salida
        if opcion == "salir":
            print("¡Gracias por usar la calculadora! Hasta pronto.")
            break

        # Ver el historial
        elif opcion == "historial":
            if not historial:
                print("\n📭 Aún no se ha realizado ninguna operación.")
            else:
                print("\n📜 Historial de operaciones:")
                for operacion in historial:
                    print(f"   • {operacion}")
            continue

        # Validamos si el usuario introdujo una operación matemática
        if opcion in ["suma", "resta", "multiplicación", "division", "división", "potencia"]:
           
            num1 = pedir_numero("Introduce el primer número: ")
            num2 = pedir_numero("Introduce el segundo número: ")

            try:
                if opcion == "suma":
                    res = sumar(num1, num2)
                    print(f"El resultado de la suma es: {res}")
                    texto_op = f"{num1} + {num2} = {res}"
                    
                elif opcion == "resta":
                    res = restar(num1, num2)
                    print(f"El resultado de la resta es: {res}")
                    texto_op = f"{num1} - {num2} = {res}"
                    
                elif opcion == "multiplicación":
                    res = multiplicar(num1, num2)
                    print(f"El resultado de la multiplicación es: {res}")
                    texto_op = f"{num1} * {num2} = {res}"
                    
                elif opcion in ["división", "division"]:
                    res = dividir(num1, num2)  # Aquí saltará error si num2 es 0
                    print(f"El resultado de la división es: {res}")
                    texto_op = f"{num1} / {num2} = {res}"
                    
                elif opcion == "potencia":
                    res = potencia(num1, num2)
                    print(f"El resultado de la potencia es: {res}")
                    texto_op = f"{num1} ** {num2} = {res}"

                # Guardamos la operación completa en el historial
                historial.append(texto_op)  

            except ValueError as e:
                print(f"❌ Error matemático: {e}")

        else:
            print("❌ Opción no reconocida. Por favor, escribe una de las opciones válidas del menú.")

#calculadora()
```
